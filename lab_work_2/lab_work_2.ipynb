```python
import pandas as pd
import numpy as np

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
avg_temp = [-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2]
precip = [38, 36, 38, 41, 53, 74, 84, 69, 47, 36, 44, 44]

df = pd.DataFrame({
    "month": months,
    "avg_temp": avg_temp,
    "precip": precip,
})

display(df.head())
print("Форма таблиці:", df.shape)

print("Індекс за замовчуванням:", df.index)

df_by_month = df.set_index("month")
display(df_by_month.loc["Тра"])

df = df_by_month.reset_index()

df.index = range(1, 13)

print("--- Вибірка через .loc (підписи від 1 до 6) ---")
display(df.loc[1:6, "avg_temp"])

print("--- Вибірка через .iloc (позиції від 0 до 5) ---")
display(df.iloc[0:6].loc[:, "avg_temp"])

print("--- Стовпці month і precip ---")
display(df.loc[:, ["month", "precip"]])

print("Місяці, де тепліше 15°C:")
display(df[df["avg_temp"] > 15]["month"])

print("\nТеплі (>10) та вологі (>50) місяці:")
display(df[(df["avg_temp"] > 10) & (df["precip"] > 50)]["month"])

print("\nДуже холодні (<0) або дуже вологі (>80) місяці:")
display(df[(df["avg_temp"] < 0) | (df["precip"] > 80)]["month"])

df["avg_temp_f"] = df["avg_temp"] * 9/5 + 32

mean_precip = df["precip"].mean()
df["is_wet"] = df["precip"] > mean_precip

def get_season(num):
    if num in [12, 1, 2]:
        return "зима"
    elif num in [3, 4, 5]:
        return "весна"
    elif num in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"

df["season"] = df.index.map(get_season)

display(df)
```
