```python
import numpy as np
from scipy import stats

# Варіант 1: Київ, Липень (генерація) = 21, стара норма mu0 = 20.0
city = "Київ"
mu0 = 20.0
gen_mean = 21.0

np.random.seed(1)
daily_temps = np.random.normal(loc=gen_mean, scale=2.5, size=30)

x_bar = daily_temps.mean()
s = daily_temps.std(ddof=1)
n = len(daily_temps)
se = s / np.sqrt(n)

print(f"Вибіркове середнє (x_bar): {x_bar:.4f}")
print(f"Вибіркове стандартне відхилення (s): {s:.4f}")
print(f"Стандартна похибка (se): {se:.4f}")

# Завдання 3: z-статистика і p-value (тест двобічний, оскільки перевіряємо зміну в будь-який бік)
z = (x_bar - mu0) / se
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print(f"z-статистика: {z:.4f}")
print(f"p-value: {p_value:.4f}")

# Завдання 5: 95% довірчий інтервал
ci_lower = x_bar - 1.96 * se
ci_upper = x_bar + 1.96 * se
print(f"95% довірчий інтервал: ({ci_lower:.4f}, {ci_upper:.4f})")
print(f"Чи потрапляє mu0 ({mu0}) в інтервал: {ci_lower <= mu0 <= ci_upper}")
```

## Відповіді на контрольні питання

1. **Що саме означає отримане p-value і чому це не «ймовірність того, що стара норма μ0 правильна»?**
   - Значення $p$-value показує ймовірність отримати таку саму або ще більш виражену відмінність вибіркового середнього від старої норми, якщо припустити, що нульова гіпотеза ($H_0$) абсолютно істинна. Це не ймовірність істинності самої гіпотези, а лише міра сумісності наявних даних із цією гіпотезою.

2. **Чому тип тесту (однобічний чи двобічний) потрібно обирати до того, як побачено результат, а не після?**
   - Вибір типу тесту (напряму гіпотези) після перегляду даних є грубим порушенням методології тестування, оскільки дослідник фактично підлаштовує критерії під уже відомий результат, що штучно завищує ймовірність помилкового висновку (хибно позитивного результату).

3. **Що в контексті даних означали б помилка I і помилка II роду?**
   - Помилка I роду означала б відхилення істинної нульової гіпотези: висновок про те, що клімат змінився і стара норма не діє, хоча насправді вона все ще правильна. Помилка II роду означала б прийняття хибної нульової гіпотези: висновок про те, що стара норма відповідає дійсності, хоча клімат на практиці вже змінився. Для кліматологів помилка I роду зазвичай є більш критичною, бо веде до необґрунтованої зміни нормативних довідників і зайвих витрат.

4. **Чи узгодився висновок тесту з висновком через довірчий інтервал?**
   - Так, висновки повністю узгодилися: оскільки $p$-value менше за рівень значущості $\alpha = 0.05$, нульова гіпотеза відхиляється, і водночас довідкове значення $\mu_0$ не потрапляє в побудований 95% довірчий інтервал. Для двобічного тесту ці два методи завжди дають математично еквівалентні рішення.
