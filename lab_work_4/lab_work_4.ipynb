```python
import numpy as np
import pandas as pd

orders = pd.DataFrame([
    {"id": 1, "місто": "Київ", "ціна": 850, "кількість": 3},
    {"id": 2, "місто": " Київ ", "ціна": "850 грн", "кількість": 2},
    {"id": 3, "місто": "КИЇВ", "ціна": 850, "кількість": np.nan},
    {"id": 4, "місто": "Kyiv", "ціна": 850, "кількість": 4},
    {"id": 5, "місто": "Київ", "ціна": 8500, "кількість": 3},
    {"id": 6, "місто": "Київ", "ціна": "850 грн", "кількість": np.nan},
    {"id": 7, "місто": "Київ", "ціна": 850, "кількість": 5},
    {"id": 7, "місто": "Київ", "ціна": 850, "кількість": 5}
])

print("--- Крок 1. Брудний набір даних ---")
display(orders)

print("\n--- Крок 2. Пропуски ---")
print("Кількість пропусків у стовпцях:\n", orders.isna().sum())
median_qty = orders["кількість"].median()
orders["кількість"].fillna(median_qty, inplace=True)
print(f"Пропуски в 'кількість' заповнено медіаною: {median_qty}")

print("\n--- Крок 3. Дублікати ---")
print("Кількість повних дублікатів (без subset):", orders.duplicated().sum())
print("Кількість дублікатів за id:", orders.duplicated(subset=["id"]).sum())
orders.drop_duplicates(inplace=True)
print("Кількість рядків після видалення дублікатів:", len(orders))

print("\n--- Крок 4. Типи даних і узгодження категорій ---")
orders["ціна"] = orders["ціна"].astype(str).str.replace(" грн", "").astype(float)

orders["місто"] = (
    orders["місто"]
    .str.strip()
    .str.lower()
    .replace({"kyiv": "київ", "київ": "київ", "київ": "київ"})
)
orders["місто"] = orders["місто"].replace("kyiv", "київ")
print("Унікальні значення міста після приведення:", orders["місто"].unique())

print("\n--- Крок 5. Викиди ---")
q1, q3 = orders["ціна"].quantile([0.25, 0.75])
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = orders[(orders["ціна"] < lower_bound) | (orders["ціна"] > upper_bound)]
print("Викиди за методом IQR:\n", outliers)

display(orders)
```
