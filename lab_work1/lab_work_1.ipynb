```python
import numpy as np
import pandas as pd

import sqlite3
conn = sqlite3.connect('coffee_shop.db')

# Дані для Варіанта 1 (Київ): середньомісячні температури від січня до грудня
temps = np.array([-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2])

# Завдання 1. Створення масивів
# Кількість днів у місяцях не високосного року (з січня по грудень)
days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])

# Номери місяців від 1 до 12
month_numbers = np.arange(1, 13)

# shape і dtype для всіх трьох масивів
print("temps — shape:", temps.shape, "| dtype:", temps.dtype)
print("days_in_month — shape:", days_in_month.shape, "| dtype:", days_in_month.dtype)
print("month_numbers — shape:", month_numbers.shape, "| dtype:", month_numbers.dtype)
```

```python
# 1. Переводимо в фарингейти без циклу
temps_f = temps * 9/5 + 32
print("Температури в °F:", temps_f)

# 2. Скільки всього днів у році було тепліше 15°C
# Множу булеву маску на дні місяця і беремо суму
warm_days_mask = temps > 15
total_warm_days = np.sum(days_in_month[warm_days_mask])
print("Всього днів у році з температурою > 15°C:", total_warm_days)

# 3. Перевірка, що сума днів дорівнює 365
print("Сума днів у році:", days_in_month.sum())
assert days_in_month.sum() == 365, "Помилка: не 365 днів!"
```

```python
# 1. Знаходжу значення й індекс найспекотнішого та найхолоднішого місяця
max_idx = temps.argmax()
min_idx = temps.argmin()

print(f"Найтепліший місяць: індекс {max_idx} (місяць №{month_numbers[max_idx]}), температура: {temps[max_idx]}°C")
print(f"Найхолодніший місяць: індекс {min_idx} (місяць №{month_numbers[min_idx]}), температура: {temps[min_idx]}°C")

# 2. Номери місяців, де температура нижче нуля
cold_months = month_numbers[temps < 0]
print("Номери місяців із температурою нижче 0°C:", cold_months)

# 3. Температури, відсортовані за зростанням, без зміни оригіналу
sorted_temps = np.sort(temps)
print("Відсортовані температури:", sorted_temps)
print("Оригінальний масив не змінився:", temps)
```

```python
# Додаю умовні сонячні години для кожного місяця
sun_hours = np.array([50, 60, 120, 180, 250, 280, 300, 290, 200, 120, 60, 40])

# Збираю в один двовимірний масив (3 рядки, 12 стовпців)
data_matrix = np.array([temps, days_in_month, sun_hours])
print("Форма матриці:", data_matrix.shape)

# Сума й середнє по кожному рядку окремо
print("Сума по рядках (axis=1):", data_matrix.sum(axis=1))
print("Середнє по рядках (axis=1):", data_matrix.mean(axis=1))

# Сума по кожному стовпцю
col_sums = data_matrix.sum(axis=0)
print("Сума по стовпцях (axis=0):", col_sums)
```

```python
# 1. Створюю масив rows форми (2, 12)
rows = np.array([temps, days_in_month])
print("Форма rows:", rows.shape)

# 2. Створюю correction форми (3,) або (2,) - у завданні сказано (2,)
correction = np.array([1, 0])
print("Форма correction:", correction.shape)

# Якщо спробувати виконати рядки нижче, буде помилка невідповідності розмірностей:
# try:
#     print(rows + correction)
# except Exception as e:
#     print("Отримана помилка:", e)

# 3. Виправляю через .reshape(2, 1)
correction_reshaped = correction.reshape(2, 1)
print("Нова форма correction:", correction_reshaped.shape)

# Тепер додавання спрацює завдяки broadcasting
result = rows + correction_reshaped
print("Результат додавання після reshape:\n", result)
```

## Відповіді на контрольні питання

1. **Чи можете ви пояснити різницю між shape == (12,) і shape == (1, 12)?**
   - `shape == (12,)` означає одновимірний масив (вектор) із 12 елементів (немає різниці між рядком і стовпцем). `shape == (1, 12)` — це двовимірний масив (матриця), який має рівно 1 рядок і 12 стовпців.

2. **Чи розумієте чому temps[temps > 15] повертає масив іншої довжини, ніж temps?**
   - Тому що булева індексація (`temps > 15`) відбирає лише ті елементи, які задовольняють умову. Оскільки не всі місяці мають температуру вище 15°С, новий масив містить менше елементів і відповідно іншу (меншу) довжину.

3. **Чи можете пояснити, що станеться, якщо додати масив форми (3,) до масиву форми (3, 12)?**
   - Виникне помилка (`ValueError: operands could not be broadcast together`), оскільки правила broadcasting у NumPy вимагають, щоб вимірювання суміщалися з кінця (праворуч ліворуч). Масив форми `(3,)` не може бути автоматично вирівняний по 12 стовпцям матриці форми `(3, 12)`.

4. **Чи можете пояснити, чому rows + correction спочатку дає помилку, а rows + correction.reshape(2, 1) — ні?**
   - Масив `rows` має форму `(2, 12)`, а початковий `correction` — `(2,)`. NumPy намагається вирівняти їх з кінця, порівнюючи розмірності 12 та 2, що викликає конфлікт. Коли ми робимо `correction.reshape(2, 1)`, його форма стає `(2, 1)`. Тепер NumPy бачить сумісні розмірності (рядок до рядка) і успішно розширює (broadcasts) єдиний стовпчик на всі 12 стовпців матриці `rows`.
