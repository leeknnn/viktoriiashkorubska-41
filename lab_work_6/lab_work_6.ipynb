```python
import numpy as np
import pandas as pd

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
kyiv_humidity = pd.Series([84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81, 85], index=months, name="Київ")

print("--- Завдання 1: Output .describe() ---")
print(kyiv_humidity.describe())

print("\n--- Завдання 2: Ручні обчислення ---")
rozmakh = kyiv_humidity.max() - kyiv_humidity.min()
var_sample = kyiv_humidity.var(ddof=1)
var_pop = kyiv_humidity.var(ddof=0)
std_val = kyiv_humidity.std(ddof=1)
q1 = kyiv_humidity.quantile(0.25)
q3 = kyiv_humidity.quantile(0.75)
iqr = q3 - q1
cv = (std_val / kyiv_humidity.mean()) * 100

print(f"Розмах: {rozmakh}")
print(f"Дисперсія (ddof=1): {var_sample:.4f}")
print(f"Дисперсія (ddof=0): {var_pop:.4f}")
print(f"Стандартне відхилення (std): {std_val:.4f}")
print(f"Q1 (25%): {q1}, Q3 (75%): {q3}, IQR: {iqr}")
print(f"Коефіцієнт варіації (CV): {cv:.2f}%")

data = {
    "Місто": ["Київ", "Львів", "Одеса", "Харків", "Дніпро", "Чернігів", "Івано-Франківськ", "Полтава", "Суми", "Ужгород"],
    "Регіон": ["Центр", "Захід", "Південь", "Схід", "Схід", "Центр", "Захід", "Центр", "Схід", "Захід"],
    "Січ": [84, 85, 82, 83, 82, 85, 86, 84, 85, 84],
    "Лют": [82, 83, 80, 80, 79, 83, 84, 81, 82, 82],
    "Бер": [78, 80, 78, 75, 74, 79, 81, 76, 77, 79],
    "Кві": [70, 73, 74, 68, 67, 71, 75, 69, 70, 73],
    "Тра": [65, 68, 71, 62, 61, 66, 70, 63, 64, 68],
    "Чер": [63, 65, 69, 60, 59, 63, 67, 61, 62, 66],
    "Лип": [61, 63, 67, 58, 57, 61, 65, 59, 60, 64],
    "Сер": [63, 65, 68, 60, 59, 63, 67, 61, 62, 66],
    "Вер": [68, 70, 72, 66, 65, 69, 72, 67, 68, 71],
    "Жов": [75, 77, 76, 73, 72, 76, 78, 74, 75, 77],
    "Лис": [81, 82, 80, 79, 78, 82, 83, 80, 81, 82],
    "Гру": [85, 86, 83, 84, 83, 86, 87, 85, 86, 85]
}

df_all = pd.DataFrame(data)
month_cols = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
df_all["Mean_Year"] = df_all[month_cols].mean(axis=1)

print("\n--- Завдання 3: Порівняння регіонів Центр і Захід ---")
center_region = df_all[df_all["Регіон"] == "Центр"]["Mean_Year"]
west_region = df_all[df_all["Регіон"] == "Захід"]["Mean_Year"]

print("Центр (Mean_Year describe):")
print(center_region.describe())
print("\nЗахід (Mean_Year describe):")
print(west_region.describe())

cv_center = (center_region.std() / center_region.mean()) * 100
cv_west = (west_region.std() / west_region.mean()) * 100
print(f"\nCV Центр: {cv_center:.2f}%, CV Захід: {cv_west:.2f}%")

lviv_humidity = pd.Series([85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86], index=months, name="Львів")
cv_lviv = (lviv_humidity.std() / lviv_humidity.mean()) * 100

print("\n--- Завдання 5: Порівняння міст ---")
print(f"CV Києва: {cv:.2f}%")
print(f"CV Львова: {cv_lviv:.2f}%")
```

## Відповіді на контрольні питання

1. **Яка міра розсіювання стоїть за кожним із рядків у виводі .describe()?**
   - Рядок std означає вибіркове стандартне відхилення (оцінку розсіювання навколо середнього з ddof=1).
   - Рядки 25% (Q1) та 75% (Q3) задають міжквартильний розмах (IQR = Q3 - Q1), який описує діапазон центральних 50% даних.
   - Рядки min та max задають розмах варіації (max - min), тобто повний діапазон усіх значень вибірки.

2. **Чому заповнення пропуску середнім штучно зменшує std?**
   - Стандартне відхилення вимірює середнє відхилення значень від центра. Якщо замість пропуску вставити точне значення середнього арифметичного, його власне відхилення дорівнює нулю. Додавання нуля у суму квадратів відхилень додає одиницю до знаменника (кількості спостережень), але не збільшує чисельник, через що підсумковий std штучно зменшується.

3. **Чому порівнювати абсолютні значення std двох груп різного масштабу без коефіцієнта варіації некоректно?**
   - Оскільки std вимірюється в тих самих абсолютних одиницях, що й самі дані, група з більшими абсолютними значеннями часто має більший std. Коефіцієнт варіації (CV = std / mean * 100%) масштабує стандартне відхилення відносно середнього значення, дозволяючи об'єктивно порівнювати відносну варіативність вибірок різного масштабу.

4. **Чи є власний розподіл симетричним і яка міра положення репрезентативніша?**
   - Значення середнього арифметичного (74.58%) та медіани 50% (72.5%) є близькими між собою, а значення опадів/вологості розподілені відносно рівномірно від літнього мінімуму до зимового максимуму. Розподіл є близьким до симетричного, тому середня величина є цілком репрезентативною мірою положення разом із медіаною.
