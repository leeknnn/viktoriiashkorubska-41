```python
import numpy as np
import pandas as pd
from scipy import stats

# Варіант 1: Київ (Регіон: Центр, Норма: 74%)
city_name = "Київ"
norma = 74.0
morning = np.array([84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81, 85])

print("--- Завдання 1: Одновибірковий t-критерій проти норми ---")
t_stat_1, p_val_1 = stats.ttest_1samp(morning, popmean=norma)
print(f"t-статистика: {t_stat_1:.4f}, p-value: {p_val_1:.4f}")

print("\n--- Завдання 2: Ручна перевірка t-статистики ---")
x_bar = morning.mean()
s = morning.std(ddof=1)
n = len(morning)
se = s / np.sqrt(n)
t_manual = (x_bar - norma) / se
print(f"Ручний t: {t_manual:.4f} (Збігається з ttest_1samp: {np.isclose(t_stat_1, t_manual)})")

print("\n--- Завдання 3: Незалежний двовибірковий критерій (Київ проти Львова - Захід) ---")
lviv_morning = np.array([85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86])

stat_lev_3, p_lev_3 = stats.levene(morning, lviv_morning)
print(f"Критерій Лівена (Київ vs Львів): stat = {stat_lev_3:.4f}, p-value = {p_lev_3:.4f}")

# Оскільки p_lev > 0.05, приймаємо рівність дисперсій (equal_var=True)
eq_var_3 = p_lev_3 >= 0.05
t_ind_3, p_ind_3 = stats.ttest_ind(morning, lviv_morning, equal_var=eq_var_3)
print(f"ttest_ind (equal_var={eq_var_3}): t = {t_ind_3:.4f}, p-value = {p_ind_3:.4f}")

print("\n--- Завдання 4: Порівняння з іншим містом (Київ проти Одеси - Південь) ---")
odesa_morning = np.array([82, 80, 78, 74, 71, 69, 67, 68, 72, 76, 80, 83])

stat_lev_4, p_lev_4 = stats.levene(morning, odesa_morning)
print(f"Критерій Лівена (Київ vs Одеса): stat = {stat_lev_4:.4f}, p-value = {p_lev_4:.4f}")

eq_var_4 = p_lev_4 >= 0.05
t_ind_4, p_ind_4 = stats.ttest_ind(morning, odesa_morning, equal_var=eq_var_4)
print(f"ttest_ind (equal_var={eq_var_4}): t = {t_ind_4:.4f}, p-value = {p_ind_4:.4f}")

print("\n--- Завдання 5: Парний t-критерій (Ранок проти Вечора) ---")
amplitude = np.array([3, 3, 4, 5, 6, 7, 7, 7, 6, 5, 4, 3])
evening = morning - amplitude

t_rel, p_rel = stats.ttest_rel(morning, evening)
print(f"Парний t-критерій (ttest_rel): t = {t_rel:.4f}, p-value = {p_rel:.4f}")
```

## Відповіді на контрольні питання

1. **Чому для перевірки середнього однієї вибірки використовується саме t-, а не z-розподіл?**
   - Розподіл Student (t-розподіл) використовується тому, що теоретична дисперсія та стандартне відхилення генеральної сукупності $\sigma$ невідомі і оцінюються за вибіркою як $s$. Оскільки $s$ сама є випадковою величиною, це вносить додаткову невизначеність у розрахунки, яку компенсує t-розподіл з важчими хвостами при обмеженому обсязі вибірки.

2. **Чому неправильний вибір equal_var у ttest_ind без перевірки критерієм Лівена дає систематично неправильний p-value?**
   - Стандартний двовибірковий t-критерій (з припущенням про рівність дисперсій) об'єднує вибіркові дисперсії двох груп в одну спільну оцінку. Якщо справжні дисперсії суттєво відрізняються (особливо при різних розмірах вибірок), таке об'єднання спотворює стандартну похибку та ступені свободи, що призводить до хибних значень t-статистики та неправильного обчислення ймовірності $p$-value.

3. **Чому парний критерій потужніший за незалежний двовибірковий на "до/після" даних?**
   - Парний t-критерій аналізує безпосередньо різницю між парними вимірюваннями для кожного об'єкта (усуваючи міжоб'єктну варіативність). Це дозволяє прибрати базовий шум та фонові розбіжності між самими місяцями чи спостереженнями, залишаючи лише чисто спрямований ефект, що робить тест значно чутливішим та потужнішим.

4. **Чи узгоджується висновок формального критерію Лівена з порівнянням стандартних відхилень s "на око"?**
   - Так, узгоджується: критерій Лівена формально перевіряє нульову гіпотезу про рівність дисперсій. Якщо вибіркові стандартні відхилення обох міст близькі за значенням, $p$-value критерію Лівена перевищує $0.05$, підтверджуючи, що спостережувані незначні розбіжності в розкиді є лише наслідком випадкових вибіркових коливань.
