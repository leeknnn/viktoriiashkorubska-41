```python
import numpy as np
from scipy import stats

np.random.seed(1)
daily_temps = np.random.normal(loc=21, scale=2.5, size=30)

mean = daily_temps.mean()
s = daily_temps.std(ddof=1)
n = len(daily_temps)
se = s / np.sqrt(n)

print(f"Середнє (mean): {mean:.4f}")
print(f"Вибіркове стандартне відхилення (s): {s:.4f}")

# Завдання 2: Довірчий інтервал для середнього (t-розподіл)
ci_mean_95 = stats.t.interval(0.95, df=n - 1, loc=mean, scale=se)
print(f"95% довірчий інтервал для середнього: {ci_mean_95}")

# Завдання 3: Довірчий інтервал для частки
p_hat = (daily_temps > 21).mean()
n_p = n * p_hat
n_1p = n * (1 - p_hat)
print(f"Частка днів вище 21°C (p_hat): {p_hat:.4f}")
print(f"n * p_hat = {n_p}, n * (1 - p_hat) = {n_1p}")

se_p = np.sqrt(p_hat * (1 - p_hat) / n)
ci_prop_95 = stats.norm.interval(0.95, loc=p_hat, scale=se_p)
print(f"95% довірчий інтервал для частки: {ci_prop_95}")

# Завдання 4: Ширина інтервалу при різних рівнях довіри
levels = [0.90, 0.95, 0.99]
print("\n--- Ширина інтервалів для середнього ---")
for level in levels:
    ci = stats.t.interval(level, df=n - 1, loc=mean, scale=se)
    width = ci[1] - ci[0]
    print(f"Рівень довіри {int(level*100)}%: інтервал {ci}, ширина = {width:.4f}")
```

1. **Яка різниця між точковою оцінкою і довірчим інтервалом?**

Довірчий інтервал будує діапазон значень навколо точкової оцінки з урахуванням ступеня невизначеності, що дає адекватніше уявлення про можливе положення істинного параметра.

2. **Чому для довірчого інтервалу середнього при невідомому $\sigma$ і невеликому $n$ використовують t-розподіл, а не нормальний?**
   - Оскільки стандартне відхилення генеральної сукупності $\sigma$ невідоме і замінюється вибірковою оцінкою $s$, виникає додаткова невизначеність. Розподіл Стьюдента (t-розподіл) має важчі хвости порівняно з нормальним розподілом, що повністю компенсує цю похибку при малих та помірних розмірах вибірки.
