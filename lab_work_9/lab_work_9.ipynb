```python
import numpy as np
import pandas as pd

np.random.seed(1)

orders_data = [
    {"вік": 22, "дохід": 18000, "кількість": 3, "місто_доставки": "Київ"},
    {"вік": 35, "дохід": 45000, "кількість": 2, "місто_доставки": "Львів"},
    {"вік": 48, "дохід": 75000, "кількість": 4, "місто_доставки": "Одеса"},
    {"вік": 29, "дохід": 30000, "кількість": 3, "місто_доставки": "Харків"},
    {"вік": 52, "дохід": 90000, "кількість": 5, "місто_доставки": "Київ"},
    {"вік": 24, "дохід": 22000, "кількість": 2, "місто_доставки": "Дніпро"},
    {"вік": 41, "дохід": 60000, "кількість": 3, "місто_доставки": "Львів"},
    {"вік": 31, "дохід": 35000, "кількість": 4, "місто_доставки": "Одеса"},
    {"вік": 60, "дохід": 120000, "кількість": 3, "місто_доставки": "Харків"},
    {"вік": 26, "дохід": 25000, "кількість": 2, "місто_доставки": "Київ"},
]

clients = pd.DataFrame(orders_data)

base_price = 850
clients["сума_замовлення"] = base_price * clients["кількість"]
clients["частка_доходу"] = clients["сума_замовлення"] / clients["дохід"]

clients["вік_cut"] = pd.cut(clients["вік"], bins=3)
clients["вік_qcut"] = pd.qcut(clients["вік"], q=3)

print("Розподіл cut:\n", clients["вік_cut"].value_counts())
print("\nРозподіл qcut:\n", clients["вік_qcut"].value_counts())

age = clients["вік"]
income = clients["дохід"]

clients["вік_minmax"] = (age - age.min()) / (age.max() - age.min())
clients["дохід_minmax"] = (income - income.min()) / (income.max() - income.min())

clients["вік_z"] = (age - age.mean()) / age.std()
clients["дохід_z"] = (income - income.mean()) / income.std()

city_dummies = pd.get_dummies(clients["місто_доставки"], prefix="місто")
clients = pd.concat([clients, city_dummies], axis=1)

print("\n--- Фінальна таблиця ---")
display(clients.head())
```

## Відповіді на контрольні питання

1. **У чому різниця між pd.cut() і pd.qcut()?**
   - Метод `pd.cut()` ділить діапазон числових значень на рівні за довжиною інтервали, через що кількість спостережень у кожній групі може значно відрізнятися. Метод `pd.qcut()` розбиває дані на квантилі так, щоб у кожній групі гарантовано була приблизно однакова кількість спостережень, але самі інтервали за довжиною будуть різними.

2. **Чому для номінальної змінної (місто) застосовують one-hot кодування, а не порядкове чи числове?**
   - Номінальні змінні не мають природного порядку чи ієрархії (наприклад, одне місто не є "більшим" за інше). Просте чисельне кодування (1, 2, 3) штучно задало б хибне відношення порядку та відстаней між категоріями, тоді як one-hot кодування створює незалежні бінарні стовпці без введення неіснуючої шкали.

3. **Навіщо масштабувати змінні перед методами, чутливими до масштабу?**
   - Чутливі до масштабу алгоритми (наприклад, методи засновані на відстанях або регуляризована регресія) оцінюють вплив змінних за їхніми числовими амплітудами. Змінна з більшим діапазоном значень (як-от дохід у десятки тисяч) домінуватиме над змінною з меншим діапазоном (вік від 18 до 70), тому масштабування приводить їх до спільної бази для рівноправного аналізу.

4. **Чому похідна змінна (наприклад, частка доходу на замовлення) часто несе більше сенсу для аналізу, ніж вихідні складники окремо?**
   - Сама по собі абсолютна сума замовлення чи рівень доходу не дають контексту про фінансове навантаження на клієнта. Похідна змінна (частка доходу) об'єднує обидва показники в один відносний індикатор, який напряму відображає купівельну спроможність та поведінкові особливості споживача незалежно від масштабу його заробітку.
